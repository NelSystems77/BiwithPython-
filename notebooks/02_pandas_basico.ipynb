# Modulo 2 · Pandas: cargar y explorar datos

**pandas** es la libreria central para el analisis de datos en Python.
En este modulo aprenderas a cargar un archivo CSV, explorar su estructura,
seleccionar columnas y filas, y hacer filtros basicos.

**Contenidos:**
1. Cargar un CSV con `read_csv`
2. Explorar la estructura: `shape`, `dtypes`, `info`, `describe`
3. Seleccionar columnas y filas
4. Filtrar filas con condiciones
5. Ordenar datos


## 1. Cargar el dataset

```python
import pandas as pd

df = pd.read_csv("../data/ventas.csv", parse_dates=["fecha"])
df.head()
```

## 2. Explorar la estructura del DataFrame

```python
print("Dimensiones (filas, columnas):", df.shape)
print()
print("Tipos de datos por columna:")
print(df.dtypes)
```

```python
df.info()
```

```python
# describe() da un resumen estadistico de las columnas numericas
df.describe()
```

```python
# Tambien podemos ver un resumen de columnas categoricas (texto)
df.describe(include="object")
```

```python
# Ver los nombres de columnas y una muestra aleatoria de filas
print(df.columns.tolist())
df.sample(5, random_state=1)
```

## 3. Seleccionar columnas y filas

```python
# Seleccionar una sola columna -> devuelve una Serie
df["producto"].head()
```

```python
# Seleccionar varias columnas -> devuelve un DataFrame
df[["fecha", "region", "producto", "ingreso"]].head()
```

```python
# Seleccionar filas por posicion con iloc
df.iloc[0:5]
```

```python
# Seleccionar filas y columnas por etiqueta con loc
df.loc[0:4, ["producto", "region", "ingreso"]]
```

## 4. Filtrar filas con condiciones

```python
# Ventas de la region Norte
ventas_norte = df[df["region"] == "Norte"]
print(f"{len(ventas_norte)} ventas en la region Norte")
ventas_norte.head()
```

```python
# Combinar condiciones: ventas de la categoria Electronica con ingreso mayor a 100.000
filtro = (df["categoria"] == "Electronica") & (df["ingreso"] > 100000)
df[filtro].head()
```

```python
# isin() para filtrar por varios valores posibles
regiones_interes = ["Norte", "Sur"]
df[df["region"].isin(regiones_interes)].shape
```

## 5. Ordenar datos

```python
# Las 10 ventas con mayor ingreso
df.sort_values("ingreso", ascending=False).head(10)[["fecha", "producto", "region", "ingreso"]]
```
