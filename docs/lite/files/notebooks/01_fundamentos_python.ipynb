## 1. Variables y tipos de datos

```python
# Tipos de datos basicos que usaremos constantemente en BI
producto = "Audifonos"          # str
precio_unitario = 15990.0       # float
unidades_vendidas = 3           # int
tiene_descuento = True          # bool

print(type(producto), type(precio_unitario), type(unidades_vendidas), type(tiene_descuento))
```

```python
# Operaciones aritmeticas basicas: calcular el ingreso de una venta
ingreso = precio_unitario * unidades_vendidas
print(f"Ingreso de la venta: ${ingreso:,.0f}")

descuento_pct = 0.10
ingreso_con_descuento = ingreso * (1 - descuento_pct)
print(f"Ingreso con descuento del {descuento_pct:.0%}: ${ingreso_con_descuento:,.0f}")
```

## 2. Estructuras de datos: listas, tuplas y diccionarios

```python
# Una lista de ventas diarias (en pesos)
ventas_semana = [125000, 98000, 143000, 87000, 210000, 305000, 276000]

print("Ventas de la semana:", ventas_semana)
print("Venta del primer dia:", ventas_semana[0])
print("Venta del ultimo dia:", ventas_semana[-1])
print("Ventas del fin de semana (viernes a domingo):", ventas_semana[4:7])
print("Total de la semana:", sum(ventas_semana))
print("Promedio diario:", sum(ventas_semana) / len(ventas_semana))
```

```python
# Un diccionario representa una venta con varios atributos (similar a una fila de una tabla)
venta = {
    "producto": "Cafetera",
    "region": "Centro",
    "unidades": 2,
    "precio_unitario": 29990,
}

ingreso = venta["unidades"] * venta["precio_unitario"]
print(f"{venta['producto']} ({venta['region']}): ${ingreso:,.0f}")
```

```python
# Una lista de diccionarios es como una mini tabla de datos
ventas = [
    {"producto": "Mouse", "region": "Norte", "unidades": 5, "precio_unitario": 6990},
    {"producto": "Teclado", "region": "Sur", "unidades": 2, "precio_unitario": 12990},
    {"producto": "Monitor", "region": "Centro", "unidades": 1, "precio_unitario": 129990},
]

for v in ventas:
    ingreso = v["unidades"] * v["precio_unitario"]
    print(f"{v['producto']:<10} {v['region']:<8} ${ingreso:>10,.0f}")
```

## 3. Condicionales

```python
# Clasificar una venta segun su monto
def clasificar_venta(ingreso):
    if ingreso >= 100000:
        return "Venta grande"
    elif ingreso >= 30000:
        return "Venta media"
    else:
        return "Venta pequena"

for v in ventas:
    ingreso = v["unidades"] * v["precio_unitario"]
    categoria = clasificar_venta(ingreso)
    print(f"{v['producto']:<10} ${ingreso:>10,.0f}  ->  {categoria}")
```

## 4. Bucles

```python
# Calcular el total de ingresos con un bucle for
total_ingresos = 0
for v in ventas:
    total_ingresos += v["unidades"] * v["precio_unitario"]

print(f"Ingreso total: ${total_ingresos:,.0f}")
```

```python
# while: simular como crece un fondo de inversion mes a mes hasta superar una meta
saldo = 500000
meta = 1000000
tasa_mensual = 0.02
meses = 0

while saldo < meta:
    saldo *= (1 + tasa_mensual)
    meses += 1

print(f"Se alcanza la meta en {meses} meses, con un saldo final de ${saldo:,.0f}")
```

## 5. Funciones

```python
def calcular_ingreso(unidades, precio_unitario, descuento_pct=0.0):
    """Calcula el ingreso neto de una venta aplicando un descuento opcional."""
    return unidades * precio_unitario * (1 - descuento_pct)


def calcular_utilidad(ingreso, costo):
    """Calcula la utilidad y el margen porcentual de una venta."""
    utilidad = ingreso - costo
    margen = utilidad / ingreso if ingreso else 0
    return utilidad, margen


ingreso = calcular_ingreso(unidades=4, precio_unitario=15990, descuento_pct=0.1)
utilidad, margen = calcular_utilidad(ingreso=ingreso, costo=ingreso * 0.6)

print(f"Ingreso: ${ingreso:,.0f}")
print(f"Utilidad: ${utilidad:,.0f} (margen {margen:.1%})")
```

## 6. Comprensiones de listas (list comprehensions)

```python
# Forma tradicional: calcular el ingreso de cada venta con un bucle
ingresos = []
for v in ventas:
    ingresos.append(v["unidades"] * v["precio_unitario"])
print(ingresos)
```

```python
# La misma logica con una list comprehension (mas idiomatico en Python)
ingresos = [v["unidades"] * v["precio_unitario"] for v in ventas]
print(ingresos)

# Con condicion: solo ingresos de ventas grandes (> 30000)
ingresos_grandes = [i for i in ingresos if i > 30000]
print("Ventas grandes:", ingresos_grandes)
```
